### TODO: Observations [1 pt]
What differences do you observe between different regularization penalties? Why do you think this is the case?

`Across the five experiments, what we observe is a clear inverse relationship between regularization coefficient and model accuracy. At the highest penalty of reg = 1.0, the network fails to learn and hovers not much better than random guessing at approximately 10.8% accuracy with erratic fluctuations. As regularization is reduced to reg = 0.1, the model still severely underfits at around 35.8% accuracy. However, when we decrease regularization further to reg = 0.01, 0.001, and 0.0001, accuracy climbs significantly to 89.6%, 92.6%, and reaches its peak at 93.4%, accompanied by smooth and stably monotonic loss and accuracy curves.`

`This behavior occurs because the optimization objective includes an L2 penalty term. When lambda, the coefficient, is excessively large (>=0.1), this weight decay term dominates the cross-entropy gradient, aggressively forcing the weights toward a low magnitude (zero) and introducing high bias. This severely restricts the neural network's representational capacity and prevents it from fitting into the non-linear decision boundaries needed to distinguish MNIST digits. Conversely, small regularization values (10^-3$ and 10^-4) provide mild weight decay without stifling the network's expressive power, allowing it to fit the dataset features and achieve strong generalization.`
